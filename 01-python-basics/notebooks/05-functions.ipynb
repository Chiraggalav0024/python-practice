```python
def check_even(n):
    if(n%2 == 0):
        print(f"{n} is an even number")
    else:
        print(f"{n} is an odd number")

n = int(input("Enter a number"))
check_even(n)
```

```python
def add (a,b):
    result = a+b
    return result

a,b = int(input("Enter first number:")), int (input("Enter second number"))
print(f"the sum of {a} and {b} is {add(a,b)}")
```
