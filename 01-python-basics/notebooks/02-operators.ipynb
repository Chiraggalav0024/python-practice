```python
#Arithmatic Operators
a = int(input("Enter first number"))
b = int(input("Enter second number"))
print(f"your first number is {a} and second number is {b}")
print(f"addition: {a+b}")
print(f"subtraction: {a-b}")
print(f"multiplication: {a*b}")
print(f"integer devision: {a//b}")
print(f"float devision: {a/b}")
```

```python
# Comparison Operators
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print(f"your first number is {a} and second number is {b}")
print(f"equal to: {a == b}")
print(f"not equal to: {a != b}")
print(f"greater than: {a > b}")
print(f"less than: {a < b}")
print(f"greater than or equal to: {a >= b}")
print(f"less than or equal to: {a <= b}")
```

```python
# Logical Operators
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print(f"your first number is {a} and second number is {b}")
print(f"a > 0 and b > 0 : {a > 0 and b > 0}")
print(f"a > 0 or b > 0 : {a > 0 or b > 0}")
print(f"not(a > b) : {not(a > b)}")
```

```python
# Assignment Operators
a = int(input("Enter a number: "))

print(f"original value of a: {a}")

a += 5
print(f"after += 5 : {a}")

a -= 2
print(f"after -= 2 : {a}")

a *= 3
print(f"after *= 3 : {a}")

a /= 2
print(f"after /= 2 : {a}")

a //= 2
print(f"after //= 2 : {a}")

a %= 3
print(f"after %= 3 : {a}")
```

```python
# Bitwise Operators
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print(f"your first number is {a} and second number is {b}")
print(f"bitwise and: {a & b}")
print(f"bitwise or: {a | b}")
print(f"bitwise xor: {a ^ b}")
print(f"bitwise not of a: {~a}")
print(f"left shift a by 1: {a << 1}")
print(f"right shift a by 1: {a >> 1}")
```

```python
# Membership Operators
text = input("Enter a word: ")
ch = input("Enter a character to check: ")

print(f"your word is {text}")
print(f"{ch} in text : {ch in text}")
print(f"{ch} not in text : {ch not in text}")
```

```python
# Identity Operators
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(f"a = {a}")
print(f"b = {b}")
print(f"c = {c}")
print(f"a is b : {a is b}")
print(f"a is c : {a is c}")
print(f"a is not b : {a is not b}")
print(f"a == b : {a == b}")
```
