```python
import random
```

```python
mix = ["chilly", "carrot", 4, 6, True, "apple"]
print(mix)
print(mix[-1])
print(mix[0:4])
print(mix[-1:-4])
```

```python
lst = [random.randint(1,100) for x in range(50)]
print(lst)
print(sorted(lst))
```

```python
lst.sort()
print(lst)
```

```python
#inserting
lst = []
lst.append(10)
lst.append(20)
lst.extend([40,32,45,56,45])
print(lst)
lst.insert(2,15)
print(lst)
```

```python
print(lst)
print(min(lst))
print(max(lst))
print(sum(lst))
```

```python
print(sorted(lst))
print(lst)
lst.sort()
print(lst)
```

```python
print(lst)
print(lst.reverse())   #produce nothing as output
print(lst)
```

```python
#deleting
lst.pop()
print(lst)
lst.pop(4)
print(lst)
lst.remove(15)
print(lst)
del lst[1]
print(lst)
del lst[1:3]
print(lst)
lst.clear()
print(lst)
```

```python
#list comprehensions
print(lst)
print(mix.count(4))
lst = [random.randint(1,100) for x in range(10)]
print(lst)

print(x*x for x in range(10))
```
